```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("C:\\Users\\sanch\\OneDrive\\Desktop\\flight-delay-intelligence\\data\\cleaned\\cleaned_flights.csv")
```

```python
df.head()
```

```python
df["Used Date"] = pd.to_datetime(df["Used Date"])
```

#### Month Feature

```python
df["Month"] = df["Used Date"].dt.month
```

```python
df[["Used Date", "Month"]].head()
```

#### Day Feature

```python
df["Day"] = df["Used Date"].dt.day
```

```python
df[["Used Date", "Day"]].head()
```

#### Weekday Feature

```python
df["Weekday"] = df["Used Date"].dt.day_name()
```

```python
df[["Used Date", "Weekday"]].head()
```

#### Total Delay

```python
df["Total Delay"] = (
    df["Departure Delay"] +
    df["Arrival Delay"]
)
```

```python
df[[
    "Departure Delay",
    "Arrival Delay",
    "Total Delay"
]].head()
```

#### Delay Category

```python
def delay_category(delay):

    if delay == 0:
        return "On Time"

    elif delay <= 10:
        return "Minor Delay"

    elif delay <= 60:
        return "Moderate Delay"

    else:
        return "Severe Delay"
```

```python
df["Delay Category"] = df["Total Delay"].apply(
    delay_category
)
```

```python
df[[
    "Total Delay",
    "Delay Category"
]].head()
```

#### Route Feature

```python
df["Route"] = (
    df["From"] +
    "-" +
    df["To"]
)
```

```python
df[[
    "From",
    "To",
    "Route"
]].head()
```

#### Peak Hour Feature

```python
def peak_hour(dep):

    dep = int(dep)

    if 600 <= dep <= 1000:
        return "Morning Peak"

    elif 1700 <= dep <= 2100:
        return "Evening Peak"

    else:
        return "Off Peak"
```

```python
df["Peak Hour"] = df["SDEP"].apply(
    peak_hour
)
```

```python
df[[
    "SDEP",
    "Peak Hour"
]].head()
```

#### Weather Severity Score

```python
df["Weather Severity"] = (
    df["weather__hourly__windspeedKmph"] * 0.3
    +
    df["weather__hourly__humidity"] * 0.2
    +
    df["weather__hourly__precipMM"] * 0.5
)
```

```python
df[[
    "weather__hourly__windspeedKmph",
    "weather__hourly__humidity",
    "weather__hourly__precipMM",
    "Weather Severity"
]].head()
```
